# Deep Learning Experiments

This notebook runs the deep learning training scripts and summarizes metrics for ANN, LSTM, and GRU models.

```python
from pathlib import Path
import subprocess, sys

project_root = Path.cwd().resolve().parents[1]
src_dir = project_root / 'src'
for script in ['train_ann.py', 'train_lstm.py', 'train_gru.py']:
    result = subprocess.run([sys.executable, str(src_dir / script)], cwd=str(project_root), capture_output=True, text=True)
    print('SCRIPT:', script, 'EXIT:', result.returncode)
    print(result.stdout.strip()[-600:])
    if result.stderr:
        print('ERROR:', result.stderr.strip()[-600:])
```
