# OULAD Data Understanding and Cleaning

This notebook validates the OULAD tables, checks data quality, and records the cleaning decisions before feature engineering.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().resolve().parents[1]
data_dir = project_root / 'data' / 'oulad'
print('project_root=', project_root)
print('data_dir=', data_dir)

required = [
    'studentInfo.csv',
    'studentRegistration.csv',
    'studentAssessment.csv',
    'studentVle.csv',
    'assessments.csv',
    'courses.csv',
    'vle.csv',
]
for name in required:
    path = data_dir / name
    print(name, path.exists(), path.stat().st_size if path.exists() else 0)
```

```python
summary = []
for name in required:
    df = pd.read_csv(data_dir / name)
    summary.append({
        'file': name,
        'rows': len(df),
        'columns': len(df.columns),
        'missing_values': int(df.isna().sum().sum()),
        'duplicate_rows': int(df.duplicated().sum()),
        'columns_list': list(df.columns),
    })
pd.DataFrame(summary)
```

## Cleaning decisions
- Preserve the original source tables and work from read-only copies.
- Convert date columns to numeric where necessary and handle missing registrations carefully.
- Keep duplicate detection explicit; do not delete rows until the meaning is understood.
- Aggregate the very large `studentVle.csv` by student-course-week rather than storing an unnecessary full copy.
