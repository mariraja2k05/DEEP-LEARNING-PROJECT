# OULAD Exploratory Data Analysis

This notebook summarizes activity, course differences, assessment patterns, and outcome distributions using the validated OULAD tables.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

project_root = Path.cwd().resolve().parents[1]
data_dir = project_root / 'data' / 'oulad'
student_info = pd.read_csv(data_dir / 'studentInfo.csv')
student_vle = pd.read_csv(data_dir / 'studentVle.csv')
student_assessment = pd.read_csv(data_dir / 'studentAssessment.csv')

figures_dir = project_root / 'outputs' / 'figures'
figures_dir.mkdir(parents=True, exist_ok=True)
sns.set_theme(style='whitegrid')

student_info['final_result'].value_counts().plot(kind='bar', figsize=(8, 4), title='Academic outcome distribution')
plt.tight_layout()
plt.savefig(figures_dir / 'outcome_distribution.png', dpi=200)
plt.show()
```

```python
weekly_clicks = student_vle.groupby(['id_student', 'code_module', 'code_presentation'], as_index=False)['sum_click'].sum()
weekly_clicks['sum_click'].describe()
plt.figure(figsize=(8, 4))
plt.hist(weekly_clicks['sum_click'], bins=40, range=(0, 5000), alpha=0.75)
plt.title('Distribution of total student clicks')
plt.xlabel('Total clicks')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig(figures_dir / 'student_click_distribution.png', dpi=200)
plt.show()
```
